```python
#### PRACTICAL 2
# Conditional Statements and Loops

# 1. Check Positive Number (if)
num = int(input("Enter a number: "))
if num > 0:
    print("Positive Number")

# 2. Even or Odd (if-else)
num = int(input("Enter another number: "))
if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")

# 3. Grade (if-elif-else)
marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
else:
    print("Fail")

# 4. Login (Nested if-else)
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":
    if password == "1234":
        print("Login Successful")
    else:
        print("Incorrect Password")
else:
    print("Invalid Username")

# 5. Voting Eligibility (One-line if-else)
age = int(input("Enter your age: "))
print("Eligible to Vote" if age >= 18 else "Not Eligible to Vote")

# 6. For Loop
print("Numbers using for loop:")
for i in range(1, 6):
    print(i)

# 7. While Loop
print("Numbers using while loop:")
i = 1
while i <= 5:
    print(i)
    i += 1

# 8. Break Statement
print("Break Example:")
for i in range(1, 11):
    if i == 5:
        break
    print(i)

# 9. Continue Statement
print("Continue Example:")
for i in range(1, 11):
    if i == 5:
        continue
    print(i)

# 10. Pass Statement
print("Pass Example:")
for i in range(1, 6):
    if i == 3:
        pass
    print(i)

# 11. Match-Case Statement
day = int(input("Enter day number (1-7): "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid Day")

print("Program Ended")
```

```python
# Practical 3 - User Defined Functions
# 1. No arguments, No return value
def greet():
    print("Hello!")
greet()
# 2. Arguments, No return value
def add(a, b):
    print("Sum =", a + b)

add(10, 20)

# 3. No arguments, Return value
def message():
    return "Welcome to Python"

print(message())

# 4. Arguments, Return value
def multiply(a, b):
    return a * b

print("Product =", multiply(5, 4))

# 5. Default argument
def student(name="varsha"):
    print("Name:", name)

student()
student("saru")

# 6. Keyword arguments
def details(name, age):
    print("Name:", name)
    print("Age:", age)

details(age=18, name="Anu")

# 7. Variable length arguments
def total(*marks):
    print("Total Marks =", sum(marks))

total(80, 90, 85)

# 8. Lambda function
square = lambda x: x * x
print("Square =", square(5))

# 9. Recursive function
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

print("Factorial =", fact(5))

# 10. Nested function
def outer():
    def inner():
        print("Inner Function")
    inner()

outer()
```

```python
# Indexing, Slicing and Reshaping in NumPy

import numpy as np

# Creating NumPy Array
array1 = np.array([10, 20, 30, 40, 50, 60])

print("Original Array:")
print(array1)

# ---------------- INDEXING ----------------

print("\nIndexing Operations")

print("First Element:", array1[0])
print("Third Element:", array1[2])
print("Last Element:", array1[-1])

# ---------------- SLICING ----------------

print("\nSlicing Operations")

print("Elements from Index 1 to 4:")
print(array1[1:5])

print("Elements from Beginning to Index 3:")
print(array1[:4])

print("Elements from Index 2 to End:")
print(array1[2:])

print("Elements with Step Size 2:")
print(array1[::2])

# ---------------- 2D ARRAY ----------------

array2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("\n2D Array:")
print(array2)

# 2D Array Indexing
print("\nElement at Row 1 Column 2:")
print(array2[0, 1])

print("Element at Row 2 Column 3:")
print(array2[1, 2])

# 2D Array Slicing
print("\nFirst Row:")
print(array2[0])

print("Second Column:")
print(array2[:, 1])

# ---------------- RESHAPING ----------------

array3 = np.array([1, 2, 3, 4, 5, 6])

print("\nOriginal 1D Array:")
print(array3)

# Reshape into 2 Rows and 3 Columns
reshaped1 = array3.reshape(2, 3)

print("\nReshaped Array (2x3):")
print(reshaped1)

# Reshape into 3 Rows and 2 Columns
reshaped2 = array3.reshape(3, 2)

print("\nReshaped Array (3x2):")
print(reshaped2)

# Flattening Array
flattened = reshaped1.flatten()

print("\nFlattened Array:")
print(flattened)

# Array Attributes
print("\nShape of Array:")
print(reshaped1.shape)

print("\nDimensions of Array:")
print(reshaped1.ndim)

print("\nSize of Array:")
print(reshaped1.size)
```

```python
#practical13
# Grouping and Aggregation Data using pandas

import pandas as pd

# Creating DataFrame
data = {
    "Department": ["HR", "IT", "HR", "IT", "Sales", "Sales"],
    "Employee": ["mathi", "saru", "Priya", "Anu", "varsha", "rithi"],
    "Salary": [30000, 50000, 35000, 60000, 45000, 40000],
    "Experience": [2, 5, 3, 6, 4, 2]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Grouping by Department
grouped = df.groupby("Department")

# Sum Aggregation
print("\nSum of Salary by Department:")
print(grouped["Salary"].sum())

# Mean Aggregation
print("\nAverage Salary by Department:")
print(grouped["Salary"].mean())

# Maximum Salary
print("\nMaximum Salary by Department:")
print(grouped["Salary"].max())

# Minimum Salary
print("\nMinimum Salary by Department:")
print(grouped["Salary"].min())

# Multiple Aggregations
print("\nMultiple Aggregations:")
print(
    grouped["Salary"].agg(
        ["sum", "mean", "max", "min"]
    )
)

# Grouping Multiple Columns
print("\nGrouping by Department and Experience:")

group_multi = df.groupby(
    ["Department", "Experience"]
)["Salary"].sum()

print(group_multi)

# Iterating Groups
print("\nDisplaying Each Group:")

for name, group in grouped:
    print("\nDepartment:", name)
    print(group)

# Sorting Values
print("\nSorted DataFrame by Salary:")

sorted_df = df.sort_values(by="Salary")
print(sorted_df)
```

```python
#practical14
# Merge and Join DataFrames using Pandas

import pandas as pd

# ----- CONCATENATE OPERATIONS ------------------- (all columns name should be same)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 1:")
print(df1)

#  Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [201, 202, 203, 205],
    "Marks": [85, 78, 80, 65]
}, index=[4, 5, 6, 7])   

print("\nDataFrame 2:")
print(df2)

frames = [df1,df2]
result = pd.concat(frames)
print("concatenate dataframes")
print(result)

# ---------------- MERGE OPERATIONS ---------------- ( at least one column should be same, others can be different  )

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["varsha", "saru", "siru", "sadhana"]
})

print("DataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 2:")
print(df2)

# Inner Merge
inner_merge = pd.merge(df1, df2, on="ID", how="inner")

print("\nInner Merge:")
print(inner_merge)

# Left Merge
left_merge = pd.merge(df1, df2, on="ID", how="left")

print("\nLeft Merge:")
print(left_merge)

# Right Merge
right_merge = pd.merge(df1, df2, on="ID", how="right")

print("\nRight Merge:")
print(right_merge)

# Outer Merge
outer_merge = pd.merge(df1, df2, on="ID", how="outer")

print("\nOuter Merge:")
print(outer_merge)

# ---------------- JOIN OPERATIONS ---------------- ( columns names in both datframes is different )

# Creating DataFrames for Join
df3 = pd.DataFrame({"jerry","mathi","kavi"}, 
index=[1, 2, 3])

df4 = pd.DataFrame({
    "Salary": [30000, 40000, 50000]
}, index=[1, 2, 3])

print("\nDataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)

# Joining DataFrames
joined_df = df3.join(df4)

print("\nJoined DataFrame:")
print(joined_df)

# Joining with Different Join Types
left_join = df3.join(df4, how="left")

print("\nLeft Join:")
print(left_join)

right_join = df3.join(df4, how="right")

print("\nRight Join:")
print(right_join)

outer_join = df3.join(df4, how="outer")

print("\nOuter Join:")
print(outer_join)


```

```python
# PRACTICAL 18
# Basic Data Cleaning and Preprocessing

import pandas as pd
import numpy as np

# Loading CSV File
file_path = (r"C:\Users\varsha\OneDrive\Desktop\weekly task\students.csv")


# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display original dataset
print("========== ORIGINAL DATASET ==========")
print(df)

# 3. Display dataset information
print("\n========== ORIGINAL INFORMATION ==========")
df.info()

# 4. Check missing values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# 5. Fill missing Age values with mean
if "Age" in df.columns:
    df["Age"] = df["Age"].fillna(df["Age"].mean())

# 6. Fill missing Marks values with mean
if "Marks" in df.columns:
    df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# 7. Fill missing City values with "Unknown"
if "City" in df.columns:
    df["City"] = df["City"].fillna("Unknown")

# 8. Remove duplicate records
df = df.drop_duplicates()

print("\n========== AFTER REMOVING DUPLICATES ==========")
print(df)

# 9. Remove extra spaces from Name column
if "Name" in df.columns:
    df["Name"] = df["Name"].str.strip()

# 10. Convert Name column to lowercase
if "Name" in df.columns:
    df["Name"] = df["Name"].str.lower()

# 11. Convert City column to lowercase
if "City" in df.columns:
    df["City"] = df["City"].str.lower()

# 12. Convert Age into integer
if "Age" in df.columns:
    df["Age"] = df["Age"].astype(int)

# 13. Rename Marks column as Score
if "Marks" in df.columns:
    df.rename(columns={"Marks": "Score"}, inplace=True)

# 14. Sort dataset by Score
if "Score" in df.columns:
    df = df.sort_values(by="Score", ascending=False)

# 15. Reset the index
df = df.reset_index(drop=True)

# 16. Display cleaned dataset
print("\n========== CLEANED DATASET ==========")
print(df)

# 17. Check missing values after cleaning
print("\n========== MISSING VALUES AFTER CLEANING ==========")
print(df.isnull().sum())

# 18. Display final dataset information
print("\n========== FINAL DATASET INFORMATION ==========")
df.info()
```

```python
# PRACTICAL 19
# Calculate Basic Statistics from a Dataset

import pandas as pd
import numpy as np

# Loading CSV File
file_path = (r"C:\Users\varsha\OneDrive\Desktop\weekly task\students.csv")


# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display the dataset
print("========== DATASET ==========")
print(df)

# 3. Select Marks column
marks = df["Marks"]

# 4. Remove missing values
marks = marks.dropna()

# 5. Count number of observations
count = marks.count()

# 6. Calculate total marks
total = marks.sum()

# 7. Calculate Mean
mean = marks.mean()

# 8. Calculate Median
median = marks.median()

# 9. Calculate Mode
mode = marks.mode()

# 10. Calculate Standard Deviation
std = marks.std()

# 11. Calculate Variance
variance = marks.var()

# 12. Find Minimum
minimum = marks.min()

# 13. Find Maximum
maximum = marks.max()

# 14. Calculate Range
data_range = maximum - minimum

# 15. Calculate First Quartile
q1 = marks.quantile(0.25)

# 16. Calculate Third Quartile
q3 = marks.quantile(0.75)

# 17. Display results
print("\n========== BASIC STATISTICS ==========")

print("Number of Observations :", count)
print("Total Marks            :", total)
print("Mean                   :", mean)
print("Median                 :", median)
print("Mode                   :", mode.tolist())
print("Standard Deviation     :", std)
print("Variance               :", variance)
print("Minimum                :", minimum)
print("Maximum                :", maximum)
print("Range                  :", data_range)
print("First Quartile (Q1)    :", q1)
print("Third Quartile (Q3)    :", q3)
```
